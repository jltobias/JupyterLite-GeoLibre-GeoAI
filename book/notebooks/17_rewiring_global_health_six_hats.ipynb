## A novice’s GeoAI translation

GeoAI is not simply a colorful map and it is not automatically an oracle. In this project, GeoAI means combining geographic representation, data engineering, spatial statistics, machine learning, visualization, and accountable interpretation. The SaTScan result is one analytical lens; the 2D map, animated scenario comparison, 3D scene, table, and six-hat review are linked ways to ask whether the result is useful and safe.

## Rewiring principles

1. **Start with a decision, not a model.** Name who needs to decide what, by when, and with which evidence.
2. **Keep the exchange boundary inspectable.** Use STAC metadata, GeoJSON, tables, and cited assets so people can reproduce the handoff.
3. **Link views, do not decorate.** A map shows where; a chart shows how much; a time slider shows change; a 3D scene shows shape; a table lets people challenge exact values.
4. **Make uncertainty and disagreement visible.** Purple synthesis is a coordination step, not a license to average away conflict.
5. **Protect people and places.** Coarsen, aggregate, govern, and review sensitive geographies before publishing.
6. **Keep humans accountable.** AI can propose, compare, summarize, and help explore; people remain responsible for consequential choices.
