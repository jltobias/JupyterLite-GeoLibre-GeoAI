### Code guide 1 — Turn perspectives into a readable table

**Read the Python:** `read_csv` opens the previous lesson's ten-row report derivative. Each tuple supplies one hat, mode and question; the column list names those fields. This is an authored reflection framework, with purple added for synthesis.

**Look for:** Seven rows of questions: six de Bono perspectives plus the project's purple extension.

**Pause and try:** Pick the white-hat question and identify a PBIX field you would need to verify.


```python
import pandas as pd
from IPython.display import display, Markdown
risk = pd.read_csv('data/pbix_top10_sat_scan.csv')
hats = pd.DataFrame([('White','Facts & evidence','Trace source fields, dates, denominators and geometry.'),('Red','Feelings & lived experience','Invite residents and practitioners to describe what the model misses.'),('Black','Risks & challenges','Test bias, false precision, privacy, missingness and harmful interpretation.'),('Yellow','Benefits & opportunities','Identify decisions where earlier, clearer spatial evidence could help.'),('Green','Creativity & alternatives','Compare baselines, local knowledge, new STAC layers and other hypotheses.'),('Blue','Process & next steps','Assign ownership, validation, review dates and stopping rules.'),('Purple','Orchestration & synthesis','Integrate perspectives without erasing disagreement.')], columns=['hat','mode','question'])
hats
```

## A novice’s GeoAI translation

GeoAI is not simply a colorful map and it is not automatically an oracle. In this project, GeoAI means combining geographic representation, data engineering, spatial statistics, machine learning, visualization, and accountable interpretation. The SaTScan result is one analytical lens; the 2D map, animated scenario comparison, 3D scene, table, and six-hat review are linked ways to ask whether the result is useful and safe.

### Code guide 2 — Build workshop prompts with Python strings

**Read the Python:** `itertuples` visits the perspectives. An f-string inserts each hat's labels into Markdown, and `join` assembles the list. `len(risk)` counts the report rows. Python formats an agenda here; it does not run seven autonomous agents or generate a model consensus.

**Look for:** A workshop question list and a case snapshot referencing ten reported clusters.

**Pause and try:** Write one competing interpretation, then assign who will check it and what evidence would change your mind.


```python
display(Markdown('### Questions to carry into the next workshop\n\n' + '\n'.join(f"- **{r.hat} Hat — {r.mode}:** {r.question}" for r in hats.itertuples())))
display(Markdown('### PBIX case snapshot\n\nThe attached report contributes **' + str(len(risk)) + '** top-ranked SaTScan clusters. The next responsible action is validation of cluster definitions, source dates, scenario semantics, local context, and intended decision with accountable partners.'))
```

## Rewiring principles

1. **Start with a decision, not a model.** Name who needs to decide what, by when, and with which evidence.
2. **Keep the exchange boundary inspectable.** Use STAC metadata, GeoJSON, tables, and cited assets so people can reproduce the handoff.
3. **Link views, do not decorate.** A map shows where; a chart shows how much; a time slider shows change; a 3D scene shows shape; a table lets people challenge exact values.
4. **Make uncertainty and disagreement visible.** Purple synthesis is a coordination step, not a license to average away conflict.
5. **Protect people and places.** Coarsen, aggregate, govern, and review sensitive geographies before publishing.
6. **Keep humans accountable.** AI can propose, compare, summarize, and help explore; people remain responsible for consequential choices.
